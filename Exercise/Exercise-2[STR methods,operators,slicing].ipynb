```python
# Indexing - It's used to access the value based on string position
# always end value will be greater than start value in positive index
# Positive Indexing -  0  1  2  3  4
# Negative Indexing - -1 -2 -3 -4 -5
# always start value will be lesser than end value in positive index

name = 'Deepak Ganesh'
print(name[0:6])
print(name[7:])
print(name[0:6:2])
print(name[-7:])
print(name[-1::-1]) # reverse a string
print(name[3])
print(name[:])
```

```python
# operators
# Arithmetic, Logical, Relational, Assignent operator

#Arithmetic
a=5
b=2
print(a+b,"Addition")
print(a-b,"Subraction")
print(a*b,"Multiplication")
print(a/b,"Division")
print(a//b,"Floor Division")
print(a%b,"Modulus(Remainder)")
print(a**b,"Power Operator like 5^2")
```

```python
# Assigment Operator

age = 5
age+=5
print(age)
age-=5
print(age)
age*=5
print(age)
age/=5
print(age)
age//=5
print(age)
age%=5
print(age)
age**=5
print(age)
```

```python
#Relative Operation

# Operator	Description
# ==		Equal
# !=		Not Equal
# >		Greater than
# <		Less than
# >=		Greater or Equal
# <=		Less or Equal

a = 5
b = 3
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

```python
#Logical Operator

# Operator	Description
# and		Both True
# or		Either True
# not		Reverse the result

a = 3
b = 5

print(a<=b and b>a)
print(a<=b and b==a)
print(b==a and a<=b)

print(a<=b or b>a)
print(a<=b or b==a)
print(b==a or a>=b)

print(not(a>b))
```
