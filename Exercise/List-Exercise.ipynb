```python
name = input('Enter the student name:')
age = input('Enter the student age:')
percentage = input('Enter the student percentage:')
status = input('Enter the student percentage:')

print(age.isdigit())

if name.isalpha() and age.isdigit() and isinstance(float(percentage),float) and (status == 'True' or status == 'False') :
    print(f'Name: {name} Age: {age} Percentage: {percentage} Is Student: {status}')
elif not name.isalpha():
    print('Please Enter the name with alphabet')
elif not age.isdigit():
    print('Please Enter the age with numbers')
elif not isinstance(float(percentage),float):
     print('Please Enter the percentage with decimal numbers')
elif status != 'True' or status !='False':
    print('Please Enter the status value as True or False')
else:
    print('Please try again later')
```

```python
from functools import reduce

marks = input('Enter the marks:')

mark_list = marks.split(',')
result = reduce(lambda a,b: int(a)+int(b), mark_list)
print(result)
```
